## String analyzer

```python
text = input("Enter a sentence: ").lower()
char_to_find = input("Enter a character to find: ").lower()

first_pos = text.find(char_to_find)

last_pos = text.rfind(char_to_find)

print(f"\nOriginal text: '{text}'")
print(f"First occurrence of '{char_to_find}': {first_pos}")
print(f"Last occurrence of '{char_to_find}': {last_pos}")

if first_pos != -1:
    substring = text[first_pos:]
    print(f"Substring from first occurrence to the end: '{substring}'")
else:
    print(f"Character '{char_to_find}' not found.")
```
